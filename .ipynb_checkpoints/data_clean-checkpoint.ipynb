```python
lineitem
```

```python
lineitem.info()
```

```python
lineitem['l_returnflag'].value_counts()
```

```python
lineitem['l_linestatus'].value_counts()
```

```python
lineitem['l_commitdate'] = pd.to_datetime(lineitem['l_commitdate'], errors='coerce')
lineitem['l_shipdate'] = pd.to_datetime(lineitem['l_shipdate'], errors='coerce')
lineitem['l_receiptdate'] = pd.to_datetime(lineitem['l_receiptdate'], errors='coerce')
```

```python
lineitem.info()
```

```python
lineitem['l_linestatus'] = lineitem['l_linestatus'].map({'F': True,'O': False
})
```

```python
lineitem
```

```python
lineitem.info()
```

```python
lineitem.to_parquet('lineitem.parquet')
```
