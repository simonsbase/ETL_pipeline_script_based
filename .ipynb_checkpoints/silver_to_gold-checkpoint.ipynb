```python
import pandas as pd
from pathlib import Path

# =========================
# 路径
# =========================
SILVER_DIR = Path("/Users/yejinheng/Desktop/Gallery/Data_Engineering/project/clear_data")
GOLD_DIR = Path("/Users/yejinheng/Desktop/Gallery/Data_Engineering/project/gold_data")
GOLD_DIR.mkdir(parents=True, exist_ok=True)

# =========================
# 读取 silver parquet
# =========================
customer = pd.read_parquet(SILVER_DIR / "customer.parquet")
orders = pd.read_parquet(SILVER_DIR / "orders.parquet")
lineitem = pd.read_parquet(SILVER_DIR / "lineitem.parquet")
part = pd.read_parquet(SILVER_DIR / "part.parquet")
supplier = pd.read_parquet(SILVER_DIR / "supplier.parquet")
partsupp = pd.read_parquet(SILVER_DIR / "partsupp.parquet")
nation = pd.read_parquet(SILVER_DIR / "nation.parquet")
region = pd.read_parquet(SILVER_DIR / "region.parquet")

# =========================
# 1. 维度映射表
# nation: ['nationKey', 'name', 'regionkey', 'comment']
# region: ['regionkey', 'name', 'comment']
# =========================
nation_lu = nation.rename(columns={
    "nationKey": "nation_key",
    "name": "nation_name",
    "regionkey": "region_key",
    "comment": "nation_comment"
}).copy()

region_lu = region.rename(columns={
    "regionkey": "region_key",
    "name": "region_name",
    "comment": "region_comment"
}).copy()

# =========================
# 2. dim_region
# =========================
dim_region = nation_lu[["nation_key", "nation_name", "region_key"]].merge(
    region_lu[["region_key", "region_name"]],
    on="region_key",
    how="left"
)

dim_region.to_parquet(GOLD_DIR / "dim_region.parquet", index=False)

# =========================
# 3. dim_customer
# customer: ['c_custkey', 'c_name', 'c_address', 'c_nationkey', 'c_phone', 'c_acctbal', 'c_mktsegment', 'c_comment']
# =========================
customer_base = customer.rename(columns={
    "c_custkey": "customer_key",
    "c_name": "customer_name",
    "c_address": "customer_address",
    "c_nationkey": "nation_key",
    "c_phone": "customer_phone",
    "c_acctbal": "customer_acctbal",
    "c_mktsegment": "market_segment",
    "c_comment": "customer_comment"
}).copy()

dim_customer = customer_base.merge(
    dim_region,
    on="nation_key",
    how="left"
)

dim_customer.to_parquet(GOLD_DIR / "dim_customer.parquet", index=False)

# =========================
# 4. dim_supplier
# supplier: ['s_suppkey', 's_name', 's_address', 's_nationkey', 's_phone', 's_acctbal', 's_comment', 's_dq_no_name', 's_dq_no_phone']
# =========================
supplier_base = supplier.rename(columns={
    "s_suppkey": "supplier_key",
    "s_name": "supplier_name",
    "s_address": "supplier_address",
    "s_nationkey": "nation_key",
    "s_phone": "supplier_phone",
    "s_acctbal": "supplier_acctbal",
    "s_comment": "supplier_comment",
    "s_dq_no_name": "dq_no_name",
    "s_dq_no_phone": "dq_no_phone"
}).copy()

dim_supplier = supplier_base.merge(
    dim_region,
    on="nation_key",
    how="left"
)

dim_supplier.to_parquet(GOLD_DIR / "dim_supplier.parquet", index=False)

# =========================
# 5. dim_part
# part: ['p_partkey', 'p_name', 'p_mfgr', 'p_brand', 'p_type', 'p_size', 'p_container', 'p_retailprice', 'p_comment']
# =========================
dim_part = part.rename(columns={
    "p_partkey": "part_key",
    "p_name": "part_name",
    "p_mfgr": "manufacturer",
    "p_brand": "brand",
    "p_type": "part_type",
    "p_size": "part_size",
    "p_container": "container",
    "p_retailprice": "retail_price",
    "p_comment": "part_comment"
}).copy()

dim_part.to_parquet(GOLD_DIR / "dim_part.parquet", index=False)

# =========================
# 6. fact_order_line
# orders: ['o_orderkey', 'o_custkey', 'o_orderstatus', 'o_totalprice', 'o_orderdate', 'o_orderpriority', 'o_clerk', 'o_shippriority', 'o_comment', 'o_dq_no_date']
# lineitem: ['l_orderkey', 'l_partkey', 'l_suppkey', 'l_linenumber', 'l_quantity', 'l_extendedprice', 'l_discount', 'l_tax', 'l_returnflag', 'l_linestatus', 'l_shipdate', 'l_commitdate', 'l_receiptdate', 'l_shipinstruct', 'l_shipmode', 'l_comment']
# =========================
lineitem_base = lineitem.rename(columns={
    "l_orderkey": "order_key",
    "l_partkey": "part_key",
    "l_suppkey": "supplier_key",
    "l_linenumber": "line_number",
    "l_quantity": "quantity",
    "l_extendedprice": "extended_price",
    "l_discount": "discount",
    "l_tax": "tax",
    "l_returnflag": "return_flag",
    "l_linestatus": "line_status",
    "l_shipdate": "ship_date",
    "l_commitdate": "commit_date",
    "l_receiptdate": "receipt_date",
    "l_shipinstruct": "ship_instruct",
    "l_shipmode": "ship_mode",
    "l_comment": "line_comment"
}).copy()

orders_base = orders.rename(columns={
    "o_orderkey": "order_key",
    "o_custkey": "customer_key",
    "o_orderstatus": "order_status",
    "o_totalprice": "order_totalprice",
    "o_orderdate": "order_date",
    "o_orderpriority": "order_priority",
    "o_clerk": "clerk",
    "o_shippriority": "ship_priority",
    "o_comment": "order_comment",
    "o_dq_no_date": "dq_no_date"
}).copy()

fact_order_line = lineitem_base.merge(
    orders_base,
    on="order_key",
    how="left"
)

fact_order_line["net_amount"] = (
    fact_order_line["extended_price"] * (1 - fact_order_line["discount"])
)

fact_order_line["charge_amount"] = (
    fact_order_line["extended_price"] * (1 - fact_order_line["discount"]) * (1 + fact_order_line["tax"])
)

fact_order_line.to_parquet(GOLD_DIR / "fact_order_line.parquet", index=False)

print("Done.")
print("dim_region:", dim_region.shape)
print("dim_customer:", dim_customer.shape)
print("dim_supplier:", dim_supplier.shape)
print("dim_part:", dim_part.shape)
print("fact_order_line:", fact_order_line.shape)
```
